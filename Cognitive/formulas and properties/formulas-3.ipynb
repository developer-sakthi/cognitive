## 1. Clock Problems

### 1.1 Mirror Image of Clock

used to find the reflected time seen in a mirror

$$
\boxed{\text{Mirror Time} = 11:60 - \text{Given Time}}
$$

$$
\begin{aligned}
\text{Given Time} & : \text{Actual time shown on the clock} \\
\text{Mirror Time} & : \text{Time seen in the mirror image}
\end{aligned}
$$

## 3. Ratios and Propostion

### 3.1 Repeated Replacement (Wine and Water)

If a vessel initially contains \(V\) litres of wine and \(d\) litres are removed and replaced with water each time, then after \(n\) such operations:

$$
\boxed{
\text{Wine Left}
=
V\left(1-\frac{d}{V}\right)^n
}
$$

$$
\begin{aligned}
V & : \text{Initial quantity of wine} \\
d & : \text{Quantity removed and replaced each time} \\
n & : \text{Number of operations}
\end{aligned}
$$
