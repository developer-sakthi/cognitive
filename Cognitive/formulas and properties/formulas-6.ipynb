## 1. Arithmetic Progression (AP)

### 1.1 General Form of an AP

An Arithmetic Progression is a sequence in which the difference between consecutive terms is constant.

$$
a,\; a+d,\; a+2d,\; a+3d,\;\ldots
$$

$$
\begin{aligned}
a & : \text{First term} \\
d & : \text{Common difference}
\end{aligned}
$$

<br/>
<br/>
<br/>
<br/>

### 1.2 nth Term of an AP

$$
\boxed{a_n = a + (n-1)d}
$$

$$
\begin{aligned}
a_n & : \text{nth term} \\
a & : \text{First term} \\
d & : \text{Common difference} \\
n & : \text{Position of the term}
\end{aligned}
$$

<br/>
<br/>
<br/>
<br/>

### 1.3 Sum of First n Terms of an AP

$$
\boxed{
S_n = \frac{n}{2}\left[2a+(n-1)d\right]
}
$$

or

$$
\boxed{
S_n = \frac{n}{2}(a+l)
}
$$

$$
\begin{aligned}
S_n & : \text{Sum of first n terms} \\
a & : \text{First term} \\
l & : \text{Last term}
\end{aligned}
$$

<br/>
<br/>
<br/>
<br/>

### 1.4 Finding the Common Difference

$$
\boxed{
d = a_n - a_{n-1}
}
$$

<br/>
<br/>
<br/>
<br/>

### 1.5 Arithmetic Mean Between Two Numbers

If \(A\) and \(B\) are two numbers, then their Arithmetic Mean is

$$
\boxed{
AM = \frac{A+B}{2}
}
$$

## 2. Geometric Progression (GP)

### 2.1 General Form of a GP

A Geometric Progression is a sequence in which the ratio between consecutive terms is constant.

$$
a,\; ar,\; ar^2,\; ar^3,\;\ldots
$$

$$
\begin{aligned}
a & : \text{First term} \\
r & : \text{Common ratio}
\end{aligned}
$$

<br/>
<br/>
<br/>
<br/>

### 2.2 nth Term of a GP

$$
\boxed{
a_n = ar^{\,n-1}
}
$$

$$
\begin{aligned}
a_n & : \text{nth term} \\
a & : \text{First term} \\
r & : \text{Common ratio}
\end{aligned}
$$

<br/>
<br/>
<br/>
<br/>

### 2.3 Sum of First n Terms of a GP

For \(r \neq 1\),

$$
\boxed{
S_n =
a\left(\frac{r^n-1}{r-1}\right)
}
$$

or

$$
\boxed{
S_n =
a\left(\frac{1-r^n}{1-r}\right)
}
$$

$$
\begin{aligned}
S_n & : \text{Sum of first n terms}
\end{aligned}
$$

<br/>
<br/>
<br/>
<br/>

### 2.4 Sum of Infinite GP

Applicable when

$$
|r| < 1
$$

$$
\boxed{
S_\infty = \frac{a}{1-r}
}
$$

$$
\begin{aligned}
S_\infty & : \text{Sum to infinity}
\end{aligned}
$$

<br/>
<br/>
<br/>
<br/>

### 2.5 Finding the Common Ratio

$$
\boxed{
r = \frac{a_n}{a_{n-1}}
}
$$

<br/>
<br/>
<br/>
<br/>

### 2.6 Geometric Mean Between Two Numbers

If \(A\) and \(B\) are two positive numbers, then their Geometric Mean is

$$
\boxed{
GM = \sqrt{AB}
}
$$
