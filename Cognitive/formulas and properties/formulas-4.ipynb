# 1. Geometry (Shapes Inscribed Inside Shapes)

### 1.1 Circle Inscribed in a Square

The diameter of the circle is equal to the side of the square.

$$
\boxed{r = \frac{a}{2}}
$$

$$
\begin{aligned}
a & : \text{Side of the square} \\
r & : \text{Radius of the inscribed circle}
\end{aligned}
$$

Area of the circle:

$$
A = \pi r^2 = \frac{\pi a^2}{4}
$$

### 1.2 Square Inscribed in a Circle

The diagonal of the square is equal to the diameter of the circle.

$$
\boxed{s = r\sqrt{2}}
$$

$$
\begin{aligned}
r & : \text{Radius of the circle} \\
s & : \text{Side of the inscribed square}
\end{aligned}
$$

Area of the square:

$$
A = s^2 = 2r^2
$$

### 1.3 Circle Inscribed in an Equilateral Triangle

The circle touches all three sides of the triangle.

$$
\boxed{r = \frac{a\sqrt{3}}{6}}
$$

$$
\begin{aligned}
a & : \text{Side of the equilateral triangle} \\
r & : \text{Radius of the inscribed circle}
\end{aligned}
$$

Area of the circle:

$$
A = \frac{\pi a^2}{12}
$$

### 1.4 Equilateral Triangle Inscribed in a Circle

All three vertices of the triangle lie on the circle.

$$
\boxed{a = r\sqrt{3}}
$$

$$
\begin{aligned}
r & : \text{Radius of the circle} \\
a & : \text{Side of the inscribed triangle}
\end{aligned}
$$

Area of the triangle:

$$
A = \frac{3\sqrt{3}}{4}r^2
$$

### 1.5 Circle Inscribed in a Right Triangle

The circle touches all three sides of the right triangle.

$$
\boxed{r = \frac{a+b-c}{2}}
$$

$$
\begin{aligned}
a & : \text{First leg of the triangle} \\
b & : \text{Second leg of the triangle} \\
c & : \text{Hypotenuse} \\
r & : \text{Inradius}
\end{aligned}
$$

Alternative formula:

$$
\boxed{r = \frac{\text{Area}}{\text{Semi-Perimeter}}}
$$
