## 2. Speed and distance

### 2.1 Conversion of units

$$
\boxed{\text{km/hr} \times \frac{5}{18} = \text{m/s}}
$$

$$
\boxed{\text{m/s} \times \frac{18}{5} = \text{km/hr}}
$$

### 2.2 Proportionality of time and speed

$$
\boxed{time \propto \frac{1}{speed}}
$$

$$
\boxed{\text{example : if speed become }\frac{2}{5}\text{ then time becomes }\frac{5}{2}}
$$

## 3. Compound Interest and Simple Interest

### 3.1 Amount (Compound Interest)

$$
\boxed{A = P \left(1 + \frac{r}{n}\right)^{nt}}
$$

$$
\begin{aligned}
A & : \text{Final amount} \\
P & : \text{Principal} \\
r & : \text{Annual interest rate (in decimal)} \\
n & : \text{Number of compounding periods per year} \\
t & : \text{Time in years}
\end{aligned}
$$

### 3.2 Compound Interest

$$
\boxed{ CI =  A - P }
$$

$$
\boxed{CI =  P \left(1 + \frac{r}{n}\right)^{nt} - P }
$$

### 3.3 EMI

$$
\boxed{EMI = \frac{P \cdot r \cdot (1 + r)^n}{(1 + r)^n - 1}}
$$

$$
\begin{aligned}
P & : \text{Principal loan amount} \\
r & : \text{Monthly interest rate (annual rate ÷ 12 ÷ 100)} \\
n & : \text{Number of monthly installments}
\end{aligned}
$$

### 3.4 SIP

use it when principal is invested every month ( or ) year

$$
\boxed{A = P \cdot \left(\frac{(1+r)^n - 1}{r}\right)}
$$

$$
\begin{aligned}
A & : \text{Final accumulated amount} \\
P & : \text{Monthly investment amount} \\
r & : \text{Monthly interest rate (annual rate ÷ 12 ÷ 100)} \\
n & : \text{Total number of monthly investments}
\end{aligned}
$$

## 4. Statistics

### 4.1 Mean of continous data

$$
\boxed{\text{Mean}=\frac{\sum fx}{\sum f}}
$$

$$
\begin{aligned}
f & : \text{Frequency} \\
x & : \text{Class mark (midpoint of class interval)} \\
fx & : \text{Product of frequency and class mark} \\
\sum fx & : \text{Sum of all } fx \\
\sum f & : \text{Total frequency}
\end{aligned}
$$

### 4.2 Median of continous data

$$
\boxed{\text{Median} = L + \left(\frac{\frac{N}{2} - F}{f}\right)\cdot h}
$$

$$
\begin{aligned}
L & : \text{Lower boundary of the median class} \\
N & : \text{Total frequency} \\
F & : \text{Cumulative frequency before median class} \\
f & : \text{Frequency of median class} \\
h & : \text{Class width}
\end{aligned}
$$

median class : first class with Cumulative frequency >= N/2

### 4.3 Variance of Continuous Data

$$
\boxed{\text{Variance }(\sigma^2)=\frac{\sum f(x-\bar{x})^2}{\sum f}}
$$

$$
\begin{aligned}
f & : \text{Frequency} \\
x & : \text{Class mark (midpoint of class interval)} \\
\bar{x} & : \text{Mean of the distribution} \\
(x-\bar{x}) & : \text{Deviation from the mean} \\
f(x-\bar{x})^2 & : \text{Frequency} \times \text{ squared deviation} \\
\sum f(x-\bar{x})^2 & : \text{Sum of all weighted squared deviations} \\
\sum f & : \text{Total frequency}
\end{aligned}
$$

### 4.4 Standard Deviation of Continuous Data

$$
\boxed{\text{Standard Deviation }(\sigma)=\sqrt{\frac{\sum f(x-\bar{x})^2}{\sum f}}}
$$

$$
\begin{aligned}
\sigma & : \text{Standard deviation} \\
\sigma^2 & : \text{Variance} \\
f & : \text{Frequency} \\
x & : \text{Class mark (midpoint of class interval)} \\
\bar{x} & : \text{Mean of the distribution} \\
\sum f(x-\bar{x})^2 & : \text{Sum of all weighted squared deviations} \\
\sum f & : \text{Total frequency}
\end{aligned}
$$

### 4.5 Mode of continuous data


$$
\boxed{\text{Mode} = L + \left(\frac{f_1 - f_0}{2f_1 - f_0 - f_2}\right)\cdot h}
$$

$$
\begin{aligned}
L & : \text{Lower boundary of the modal class} \\
f_1 & : \text{Frequency of the modal class} \\
f_0 & : \text{Frequency of the class preceding the modal class} \\
f_2 & : \text{Frequency of the class succeeding the modal class} \\
h & : \text{Class width}
\end{aligned}
$$

The **modal class** is the class interval having the highest frequency.


### 4.6 Empirical Relationship Between Mean, Median and Mode

$$
\boxed{\text{Mode} = 3 \times \text{Median} - 2 \times \text{Mean}}
$$

or equivalently,

$$
\boxed{\text{Mean} - \text{Mode} = 3(\text{Mean} - \text{Median})}
$$

This formula is useful when any two of Mean, Median, and Mode are known.

### 4.7 Property of Mean under Addition/Subtraction

If the same number is added to or subtracted from every observation, the mean also changes by the same number.

$$
\boxed{\text{New Mean}=\text{Old Mean}\pm k}
$$

$$
\begin{aligned}
\text{Old Mean} & : \text{Mean of original data} \\
k & : \text{Number added or subtracted to each observation} \\
+ & : \text{Used when } k \text{ is added to each observation} \\
- & : \text{Used when } k \text{ is subtracted from each observation} \\
\text{New Mean} & : \text{Mean after transformation}
\end{aligned}
$$

For the given question:

$$
\text{Old Mean}=5,\qquad k=10
$$

Since 10 is subtracted from each observation,

$$
\text{New Mean}=5-10=-5
$$

### 4.8 Coefficient of Variation (Discrete Data)

Coefficient of Variation (C.V.) is used to measure the relative variability of a data set.

$$
\boxed{\text{C.V.} = \frac{\sigma}{\bar{x}} \times 100}
$$

$$
\begin{aligned}
\sigma & : \text{Standard Deviation} \\
\bar{x} & : \text{Mean}
\end{aligned}
$$
