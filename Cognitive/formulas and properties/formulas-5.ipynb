# 1. Geometry ( SUrface Area and volumes )

### 1.1 Cube

Let \(a\) be the side of the cube.

$$
\boxed{V = a^3}
$$

$$
\boxed{\text{TSA} = 6a^2}
$$

$$
\boxed{\text{LSA} = 4a^2}
$$

$$
\begin{aligned}
a & : \text{Side of the cube} \\
V & : \text{Volume} \\
\text{TSA} & : \text{Total Surface Area} \\
\text{LSA} & : \text{Lateral Surface Area}
\end{aligned}
$$

<br/>
<br/>



### 1.2 Cuboid

Let \(l\), \(b\), and \(h\) be the length, breadth, and height respectively.

$$
\boxed{V = lbh}
$$

$$
\boxed{\text{TSA} = 2(lb + bh + hl)}
$$

$$
\boxed{\text{LSA} = 2h(l+b)}
$$

$$
\begin{aligned}
l & : \text{Length} \\
b & : \text{Breadth} \\
h & : \text{Height}
\end{aligned}
$$

<br/>
<br/>




### 1.3 Cylinder

Let \(r\) be the radius and \(h\) be the height.

$$
\boxed{V = \pi r^2 h}
$$

$$
\boxed{\text{CSA} = LSA = 2\pi rh}
$$

$$
\boxed{\text{TSA} = 2\pi r(h+r)}
$$

$$
\begin{aligned}
r & : \text{Radius of the base} \\
h & : \text{Height} \\
V & : \text{Volume} \\
\text{CSA} & : \text{Curved Surface Area} \\
\text{TSA} & : \text{Total Surface Area}
\end{aligned}
$$

<br/>
<br/>



### 1.4 Cone

Let \(r\) be the radius, \(h\) be the height, and \(l\) be the slant height.

$$
\boxed{l = \sqrt{r^2+h^2}}
$$

$$
\boxed{V = \frac{1}{3}\pi r^2 h}
$$

$$
\boxed{\text{CSA} = \pi rl}
$$

$$
\boxed{\text{TSA} = \pi r(l+r)}
$$

$$
\begin{aligned}
r & : \text{Radius of the base} \\
h & : \text{Height} \\
l & : \text{Slant Height}
\end{aligned}
$$

<br/>
<br/>




### 1.5 Sphere

Let \(r\) be the radius of the sphere.

$$`
\boxed{V = \frac{4}{3}\pi r^3}
$$

$$
\boxed{\text{SA} = 4\pi r^2}
$$

$$
\begin{aligned}
r & : \text{Radius of the sphere} \\
V & : \text{Volume} \\
\text{SA} & : \text{Surface Area}
\end{aligned}
$$
