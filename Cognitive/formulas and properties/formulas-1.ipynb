## 1. Number System 


### 1.1 Base property 

$$
\boxed{91_x = 73_{10}}
$$

$$
\boxed{9x + 1 = 73_{10}}
$$

$$
\begin{aligned}
91_x & : \text{Number }91\text{ in base }x \\[4pt]
73_{10} & : \text{Equivalent value in decimal (base 10)} \\[4pt]
9x   & : \text{Value of }9\text{ in the }x^1\text{ place} \\[4pt]
1    & : \text{Value of }1\text{ in the }x^0\text{ place} \\[4pt]
x    & : \text{Base of the number system}
\end{aligned}
$$

General property:

$$
\boxed{abc_x = (a x^2 + b x + c)_{10}}
$$

where

$$
\begin{aligned}
a,b,c & : \text{Digits of the number} \\[4pt]
x     & : \text{Base of the number system}
\end{aligned}
$$

## 2. Partnership ( Profit sharing in buisness )


### 2.1 Properties of Investment and time'

$$
\boxed{\text{Profit share} \propto \text{Investment} \times \text{Time}}
$$

$$
\text{Money-time contribution} = \text{Capital} \times \text{Duration}
$$

$$
\text{Total contribution of a person} = \sum (\text{each investment} \times \text{time it remained invested})
$$
