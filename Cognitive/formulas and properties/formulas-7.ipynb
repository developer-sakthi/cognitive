# 1. Basic Trigonometry

### 1.1 Trigonometric Ratios in a Right-Angled Triangle

For an angle $\theta$:

$$
\boxed{\sin\theta=\frac{\text{Perpendicular}}{\text{Hypotenuse}}}
$$

$$
\boxed{\cos\theta=\frac{\text{Base}}{\text{Hypotenuse}}}
$$

$$
\boxed{\tan\theta=\frac{\text{Perpendicular}}{\text{Base}}}
$$

$$
\boxed{\cot\theta=\frac{\text{Base}}{\text{Perpendicular}}}
$$

$$
\boxed{\sec\theta=\frac{\text{Hypotenuse}}{\text{Base}}}
$$

$$
\boxed{\cosec\theta=\frac{\text{Hypotenuse}}{\text{Perpendicular}}}
$$

### 1.2 Reciprocal Identities

$$
\boxed{\sin\theta=\frac{1}{\cosec\theta}}
$$

$$
\boxed{\cos\theta=\frac{1}{\sec\theta}}
$$

$$
\boxed{\tan\theta=\frac{1}{\cot\theta}}
$$

$$
\boxed{\cosec\theta=\frac{1}{\sin\theta}}
$$

$$
\boxed{\sec\theta=\frac{1}{\cos\theta}}
$$

$$
\boxed{\cot\theta=\frac{1}{\tan\theta}}
$$

<br/>
<br/>
<br/>
<br/>

### 1.3 Quotient Identities

$$
\boxed{\tan\theta=\frac{\sin\theta}{\cos\theta}}
$$

$$
\boxed{\cot\theta=\frac{\cos\theta}{\sin\theta}}
$$

<br/>
<br/>
<br/>
<br/>

### 1.4 Pythagorean Identities

$$
\boxed{\sin^2\theta+\cos^2\theta=1}
$$

$$
\boxed{1+\tan^2\theta=\sec^2\theta}
$$

$$
\boxed{1+\cot^2\theta=\cosec^2\theta}
$$

<br/>
<br/>
<br/>
<br/>

### 1.5 Standard Trigonometric Values

| $\theta$ | $0^\circ$ | $30^\circ$ | $45^\circ$ | $60^\circ$ | $90^\circ$ |
|-----------|-----------|------------|------------|------------|------------|
| $\sin\theta$ | $0$ | $\frac{1}{2}$ | $\frac{\sqrt{2}}{2}$ | $\frac{\sqrt{3}}{2}$ | $1$ |
| $\cos\theta$ | $1$ | $\frac{\sqrt{3}}{2}$ | $\frac{\sqrt{2}}{2}$ | $\frac{1}{2}$ | $0$ |
| $\tan\theta$ | $0$ | $\frac{1}{\sqrt{3}}$ | $1$ | $\sqrt{3}$ | Undefined |

### 1.6 Complementary Angle Identities

$$
\boxed{\sin(90^\circ-\theta)=\cos\theta}
$$

$$
\boxed{\cos(90^\circ-\theta)=\sin\theta}
$$

$$
\boxed{\tan(90^\circ-\theta)=\cot\theta}
$$

$$
\boxed{\cot(90^\circ-\theta)=\tan\theta}
$$

$$
\boxed{\sec(90^\circ-\theta)=\csc\theta}
$$

$$
\boxed{\csc(90^\circ-\theta)=\sec\theta}
$$

### 1.7 Pythagoras Theorem

$$
\boxed{H^2=P^2+B^2}
$$

$$
\begin{aligned}
H & : \text{Hypotenuse} \\
P & : \text{Perpendicular} \\
B & : \text{Base}
\end{aligned}
$$
